### AutoGrad 

```python
import torch
import torch.nn as nn

import matplotlib.pyplot as plt
import math
```

##### Let's create an input tensor which has evenly spaced values of the interval(0, 2) and specify requires_grad = True. (Like most functions that create tensors, torch.linspace(), accepts an optional requires_grad option)

```python
a = torch.linspace(0, 2 *  math.pi, steps = 25, requires_grad = True)
print(a)
```

```python
# detach() is useful when you want to perform operations on tensor but you do not want those operations to be part of the gradient computation 
a.detach()
```

### Let's perform a computation, and plot it's output in terms of service 

```python
#performing a sin operation to our tensor
b = torch.sin(a)
plt.plot(a.detach(), b.detach())
```

```python
print(b)
```

```python
#performing multiplication operation 
c = 2 * b #sin(x)

#performing addition operation 
d = c + 1 
print(d)

```

##### Finally, we compute a single-element output. Since when you call .backward() on a tensor with no arguments, it expects the calling tensor to contain only a single element, as is the case when computing a loss function

```python
out = d.sum()
print(out)
```

```python
print('d :')
print(d.grad_fn)
print("1")
print(d.grad_fn.next_functions)
print("2")
print(d.grad_fn.next_functions[0][0].next_functions)
print("3")
print(d.grad_fn.next_functions[0][0].next_functions[0][0].next_functions)
print("4")
print(d.grad_fn.next_functions[0][0].next_functions[0][0].next_functions[0][0].next_functions)
print('/nc:')
print(c.grad_fn)
print('\nb')
print(b.grad_fn)
print('\na:')
print(a.grad_fn)
```

```python
#we call the backward method on the output and then check the input's grad property to inspect the gradients 
out.backward()
print(a.grad)
plt.plot(a.detach(), a.grad.detach())
```
